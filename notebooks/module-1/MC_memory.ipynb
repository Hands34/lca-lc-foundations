## No Memory

```python
from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano"
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Mitchell and my favorite color is red.")

response = agent.invoke(
    {"messages": [question]}
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What is my favorite color?")

response = agent.invoke(
    {"messages": [question]}
)

pprint(response)
```

## Memory

```python
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
    model="gpt-5-nano",
    checkpointer=InMemorySaver(),
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Mitchell and my favorite color is red.")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What is my favorite color?")

response = agent.invoke(
    {"messages": [question]},
    config,
)  

pprint(response)
```
