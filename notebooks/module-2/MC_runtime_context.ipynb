```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from dataclasses import dataclass

@dataclass
class ColourContext:
    fav_colour: str = "blue"
    least_fav_colour: str = "yellow"
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano",
    context_schema=ColourContext
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favourite colour?")]},
    context=ColourContext()
)
```

```python
from pprint import pprint

pprint(response)
```

## Accessing Context

```python
from langchain.tools import tool, ToolRuntime

@tool
def get_fav_colour(runtime: ToolRuntime) -> str:
    """Get the favourite colour of the user"""
    return runtime.context.fav_colour

@tool
def get_least_fav_color(runtime: ToolRuntime) -> str:
    """Get the least favourite colour of the user"""
    return runtime.context.least_fav_colour
```

```python
agent = create_agent(
    model="gpt-5-nano",
    tools=[get_fav_colour, get_least_fav_color],
    context_schema=ColourContext
)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favourite colour?")]},
    context=ColourContext()
)

pprint(response)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favourite colour?")]},
    context=ColourContext(fav_colour="green")
)

pprint(response)
```
